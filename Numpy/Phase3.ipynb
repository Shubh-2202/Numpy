## Advance Data Manipulations:

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("====Zomato Sales Analysis====")
print("\n Sales data shape: ", sales_data.shape)

print("First 3 Sales: ", sales_data[:3, 1:])
```

```python
# print(np.sum(sales_data,axis=0))
yearly_total= np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)
print(np.sum(yearly_total))
```

```python
# min sales
min_sales= np.min(sales_data[:5 , 1:], axis=1)
print(min_sales)
```

```python
#max sales by all Restaurants

max_sales= np.max(sales_data[:5 , 1:], axis=1)
print(max_sales)
```

```python
#Total sale by each restaurant every year
tot_sales= np.sum(sales_data[:5, 1:], axis=1)
print(tot_sales)
```

```python
#Avreage sales Per year 
avg_sale= np.mean(sales_data[:5, 1:], axis=1)
print(avg_sale)
```

```python
cumsum= np.cumsum(sales_data[:5, 1:], axis=1)
print(cumsum)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumsum along restaurant!")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(False)
plt.show()
```
